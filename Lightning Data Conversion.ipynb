```python
import xarray as xr
import pandas as pd

def nc_to_csv(ncfile, filename, low_lat, high_lat, low_long, high_long):
    
    ds = xr.open_dataset(ncfile, engine="netcdf4")
    subset = ds.where((ds.lat > low_lat) & (ds.lat < high_lat) & (ds.lon > low_long) & (ds.lon < high_long), drop=True)

    bham_datap = subset.to_dataframe()
    csvfile = bham_datap.to_csv(filename, index=False)
    
    return csvfile, filename

test = nc_to_csv("WWLLN_th_2025.nc", '2025_bham_subset.csv', 52, 53, 1.5, 2.5)
```

```python
def extract_lightning(data_file):
    file = np.loadtxt(data_file, skiprows=1, usecols=range(4), delimiter = ',')
    lightning = file[:,3]
    monthly_lightning = []
    for i in range(12):
        monthly_lightning.append(int(np.sum(lightning[i::12])))
    return np.array(monthly_lightning)
```
